¿Cuál es el valor de la señal discreta $x[n]$ en $n=1$?


```python
import numpy as np
import sys
import os

from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, HoverTool

sys.path.append(os.path.abspath(".."))
from utils.plot_helpers import style_math_axes, add_math_ticks

output_notebook(verbose=False, hide_banner=True);

# ==============================================================================
# 1. SEÑAL DISCRETA
# ==============================================================================

n = np.array([0, 1, 2])
x = np.array([1, 3, 5])

source = ColumnDataSource(data=dict(
    n=n,
    x=x,
    zeros=np.zeros_like(x)
))

# ==============================================================================
# 2. FIGURA
# ==============================================================================

p = figure(
    height=350,
    width=550,
    title="Señal discreta",
    tools="pan,wheel_zoom,reset"
)

# Líneas verticales (stem plot)
p.segment(
    x0="n", y0="zeros",
    x1="n", y1="x",
    source=source,
    line_width=3,
    color="blue"
)

# Muestras discretas
markers = p.scatter(
    "n", "x",
    source=source,
    size=12,
    color="blue",
    hover_color="red",
    hover_alpha=1
)

# Mostrar valores al pasar el ratón
hover = HoverTool(
    renderers=[markers],
    tooltips=[
        ("n", "@n"),
        ("x[n]", "@x")
    ],
    mode="mouse"
)

p.add_tools(hover)

# Estilo matemático de los ejes
style_math_axes(
    p,
    x_range=(-0.5, 2.5),
    y_range=(-0.5, 6),
    xlabel="n",
    ylabel="x[n]"
)

# Marcas enteras en los ejes
add_math_ticks(
    p,
    xticks=[0, 1, 2],
    xtick_labels=["0", "1", "2"],
    yticks=[1, 2, 3, 4, 5],
    ytick_labels=["1", "2", "3", "4", "5"],
    tick_len=5
)

p.grid.grid_line_alpha = 0.25

show(p)
```

```python
import numpy as np
import sys
import os

from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, HoverTool

sys.path.append(os.path.abspath(".."))
from utils.plot_helpers import style_math_axes, add_math_ticks

output_notebook(verbose=False, hide_banner=True);

# ==============================================================================
# 1. SEÑAL DISCRETA
# ==============================================================================

n = np.array([0, 1, 2, 3, 4])
x = np.array([1, 2, 3, 4, 5])

source = ColumnDataSource(data=dict(
    n=n,
    x=x,
    zeros=np.zeros_like(x)
))

# ==============================================================================
# 2. FIGURA
# ==============================================================================

p = figure(
    height=350,
    width=550,
    title="Señal discreta",
    tools="pan,wheel_zoom,reset"
)

# Líneas verticales (stem plot)
p.segment(
    x0="n", y0="zeros",
    x1="n", y1="x",
    source=source,
    line_width=3,
    color="blue"
)

# Muestras discretas
markers = p.scatter(
    "n", "x",
    source=source,
    size=12,
    color="blue",
    hover_color="red",
    hover_alpha=1
)

# Mostrar valores al pasar el ratón
hover = HoverTool(
    renderers=[markers],
    tooltips=[
        ("n", "@n"),
        ("x[n]", "@x")
    ],
    mode="mouse"
)

p.add_tools(hover)

# Estilo matemático de los ejes
style_math_axes(
    p,
    x_range=(-0.5, 4.5),
    y_range=(-0.5, 6),
    xlabel="n",
    ylabel="x[n]"
)

# Marcas enteras en los ejes
add_math_ticks(
    p,
    xticks=[0, 1, 2, 3, 4, 5],
    xtick_labels=["0", "1", "2", "3", "4"],
    yticks=[1, 2, 3, 4, 5],
    ytick_labels=["1", "2", "3", "4", "5"],
    tick_len=5
)

p.grid.grid_line_alpha = 0.25

show(p)
```
