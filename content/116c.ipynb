## Preguntas gráficas

En los siguientes ejercicios pasa el cursor sobre las curvas o las muestras para consultar las coordenadas. En las señales discretas, **cada punto corresponde a un valor entero de $n$**. Puedes utilizar las herramientas de zoom y restablecer la vista desde la barra lateral.


```python
# Recursos de visualización 
import sys
import os
import numpy as np

sys.path.append(os.path.abspath(".."))
from utils.plot_helpers import style_math_axes, add_math_ticks
from utils.test_helpers import (
    create_answer_selector, create_numeric_input_exercise, create_true_false_exercise
)

from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource, HoverTool, CustomJS, Slider
from bokeh.layouts import column

output_notebook(verbose=False, hide_banner=True)

AZUL = "#2563eb"
ROJO = "#dc2626"
VERDE = "#059669"


def grafica_discreta(n, x, titulo, xlim, ylim, xticks, yticks,
                     ylabel="x[n]", color=AZUL):
    """Gráfica de tallos con valores en hover y ejes del libro."""
    n = np.asarray(n, dtype=int)
    x = np.asarray(x, dtype=float)
    source = ColumnDataSource(dict(n=n, x=x, zero=np.zeros(len(n))))
    p = figure(width=620, height=330, title=titulo,
               tools="pan,wheel_zoom,reset")
    p.segment("n", "zero", "n", "x", source=source,
              color=color, line_width=2.5)
    puntos = p.scatter("n", "x", source=source, size=11, color=color,
                       hover_color=ROJO)
    p.add_tools(HoverTool(renderers=[puntos], mode="mouse",
                          tooltips=[("n", "@n"), ("x[n]", "@x{0.000}")]))
    style_math_axes(p, x_range=xlim, y_range=ylim, xlabel="n", ylabel=ylabel)
    add_math_ticks(p, xticks=list(xticks), xtick_labels=[str(k) for k in xticks],
                   yticks=list(yticks), ytick_labels=[str(k) for k in yticks],
                   tick_len=6)
    return p


def grafica_continua(t, x, titulo, xlim, ylim, xticks, yticks,
                      ylabel="x(t)", color=AZUL):
    """Gráfica continua con cursor de coordenadas y ejes matemáticos."""
    source = ColumnDataSource(dict(t=np.asarray(t), x=np.asarray(x)))
    p = figure(width=620, height=330, title=titulo,
               tools="pan,wheel_zoom,reset")
    curva = p.line("t", "x", source=source, line_width=3, color=color)
    p.add_tools(HoverTool(renderers=[curva], mode="vline",
                          tooltips=[("t", "@t{0.00}"), ("x(t)", "@x{0.000}")]))
    style_math_axes(p, x_range=xlim, y_range=ylim, xlabel="t", ylabel=ylabel)
    add_math_ticks(p, xticks=list(xticks), xtick_labels=[str(k) for k in xticks],
                   yticks=list(yticks), ytick_labels=[str(k) for k in yticks],
                   tick_len=6)
    return p
```

### 7. Aproximación gaussiana del impulso de Dirac

Mueve el control de anchura $\sigma$ y observa cómo cambia la gráfica de

$$\delta_\sigma(t)=\frac{1}{\sigma\sqrt{2\pi}} e^{-t^2/(2\sigma^2)}.$$

¿Qué sucede con el **área total** bajo la curva cuando disminuye $\sigma$?

A. Tiende a cero  
B. Permanece igual a 1  
C. Crece sin límite  
D. Cambia de signo

```python
# Adaptación del ejemplo interactivo de la aproximación gaussiana de 113b.
t = np.linspace(-3, 3, 1601)
sigma0 = 0.50

def delta_gaussiana(t, sigma):
    return np.exp(-t**2 / (2 * sigma**2)) / (sigma * np.sqrt(2 * np.pi))

source = ColumnDataSource(dict(t=t, y=delta_gaussiana(t, sigma0)))
p = figure(width=620, height=345, title="Aproximación gaussiana de δ(t)",
           tools="pan,wheel_zoom,reset")
linea = p.line("t", "y", source=source, line_width=3, color=AZUL)
p.add_tools(HoverTool(renderers=[linea], mode="vline",
                      tooltips=[("t", "@t{0.00}"), ("δσ(t)", "@y{0.000}")]))
style_math_axes(p, x_range=(-3.0, 3.0), y_range=(-0.15, 2.9),
                xlabel="t", ylabel="δσ(t)")
add_math_ticks(p, xticks=[-2, -1, 0, 1, 2],
               xtick_labels=["-2", "-1", "0", "1", "2"],
               yticks=[1, 2], ytick_labels=["1", "2"], tick_len=6)
slider = Slider(start=0.15, end=1.0, value=sigma0, step=0.05,
                title="Anchura σ", width=610)
slider.js_on_change("value", CustomJS(args=dict(source=source), code="""
    const sigma = cb_obj.value;
    const t = source.data.t;
    source.data.y = t.map(v => Math.exp(-v*v/(2*sigma*sigma)) /
                               (sigma*Math.sqrt(2*Math.PI)));
    source.change.emit();
"""))
show(column(p, slider))
create_answer_selector(num_options=4, solution=2, mode="alphabetic")
```

### 15. Ambigüedad de frecuencia en tiempo discreto

Dos cosenos discretos con frecuencias $\Omega_1=\pi/4$ y $\Omega_2=9\pi/4$ están representados mediante círculos y rombos. ¿Qué afirmación es correcta?

A. Sus muestras son distintas  
B. Coinciden solo para $n=0$  
C. Son la misma secuencia para cualquier $n$ entero  
D. Las dos señales no son periódicas

```python
# Adaptación del ejemplo de ambigüedad de frecuencia de 113a.
n = np.arange(-8, 9)
omega1, omega2 = np.pi/4, 9*np.pi/4
x1, x2 = np.cos(omega1*n), np.cos(omega2*n)
s = ColumnDataSource(dict(n=n, x1=x1, x2=x2, zero=np.zeros(len(n))))
p = figure(width=620, height=350, title="Muestreo discreto de dos cosenos",
           tools="pan,wheel_zoom,reset")
p.segment("n", "zero", "n", "x1", source=s, line_width=2, color=AZUL)
a = p.scatter("n", "x1", source=s, size=11, color=AZUL,
              legend_label="Ω₁ = π/4")
b = p.scatter("n", "x2", source=s, marker="diamond", size=15,
              color=ROJO, fill_alpha=0.10, line_width=2,
              legend_label="Ω₂ = 9π/4")
p.add_tools(HoverTool(renderers=[a, b],
                      tooltips=[("n", "@n"), ("cos(Ω₁ n)", "@x1{0.000}"),
                                ("cos(Ω₂ n)", "@x2{0.000}")]))
style_math_axes(p, x_range=(-8.5, 8.5), y_range=(-1.4, 1.4),
                xlabel="n", ylabel="amplitud")
add_math_ticks(p, xticks=[-8, -4, 0, 4, 8],
               xtick_labels=["-8", "-4", "0", "4", "8"],
               yticks=[-1, 1], ytick_labels=["-1", "1"], tick_len=6)
p.legend.location = "top_left"
p.legend.click_policy = "hide"
show(p)
create_answer_selector(num_options=4, solution=3, mode="alphabetic")
```
